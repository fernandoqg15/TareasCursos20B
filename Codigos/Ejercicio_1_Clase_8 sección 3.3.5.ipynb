(a). ¿Cualés son las componentes de mi tensor bivariable dadas las bases de monomios?

```python
x, y = symbols('x, y')
a_0 , a_1 , a_2 , b_0, b_1 , b_2 = symbols('a_0 , a_1 , a_2 , b_0, b_1 , b_2')
```

```python
base_x = [x**i for i in range(3)]
base_y = [y**i for i in range(3)]
base_T = tensorproduct(base_x , base_y)
base_T
```

```python
p_x = a_0 + a_1*x + a_2*x**2
p_y = b_0 + b_1*y + b_2*y**2

producto = expand(p_x * p_y)
producto
```

```python
producto_cij = producto

a = IndexedBase('a')
b = IndexedBase('b')
c = IndexedBase('c')

for i in range(3):
    for j in range(3):
        producto_cij = producto_cij.subs(a[i] * b[j], c[i,j])

producto_cij
```

```python
coeff = Poly (producto_cij, x, y)
Matriz_cij = Matrix([
    [coeff.coeff_monomial(x**i * y**j) for j in range(3)]
    for i in range(3)
])
Matriz_cij
```

(b). Considerando el vector p(x) y g(y), expreselo en base de los polinomios de Legendre

```python
p_x0 = 3 + x + x**2
g_y0 = 1 + y

producto = expand(p_x0 * g_y0)
producto
```

```python
coeffs = Poly(producto, x, y)
Matriz_coeffs = Matrix([
    [coeffs.coeff_monomial(x**i * y**j) for j in range(3)]
    for i in range(3)
])

Matriz_coeffs
```

(c). Expresar a p_x0 en términos de la base de Legendre

```python
#Creando nuevamente nuestra base ortogonal:

a, b = -1,1

def producto_interno (f, g):
    h = simplify(f*g)
    return integrate(h, (x, a, b))

base_legendre = []

for v in base_x:
    u = v
    for o in base_legendre:
        u -= producto_interno(u, o)/producto_interno(o,o)*o
    base_legendre.append(u)

#Calcular los coeficientes:

coeficientes = []

for q_i in base_legendre:
    c_i = producto_interno(p_x0, q_i)/producto_interno(q_i , q_i)
    coeficientes.append(c_i)

for i, c in enumerate(coeficientes):
    print(f"c_{i} = {c}")
```

(c). Armamos una base de los polinomios de Legendre pero para y

```python
def producto_interno_1 (f, g):
    h = simplify(f*g)
    return integrate(h, (y, a , b))

base_legendre_y = []

for w in base_y :
    v = w
    for o in base_legendre_y:
        v -= producto_interno_1(v, o)/producto_interno_1(o, o) * o

    base_legendre_y.append(v)

```

```python
coeficientes_x = []
for q_i in base_legendre:
    num = producto_interno(p_x, q_i)
    den = producto_interno(q_i, q_i)
    c_ix = num / den
    coeficientes_x.append(c_ix)
    
coeficientes_y = []
for q_i in base_legendre_y:
    num = producto_interno_1(p_y, q_i)
    den = producto_interno_1(q_i, q_i)
    c_iy = num / den
    coeficientes_y.append(c_iy)

coef_tensor = Matrix([a * b for a in coeficientes_x for b in coeficientes_y])

print(coef_tensor)
```
